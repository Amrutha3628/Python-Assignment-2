## List Creation

```python
age_list= [24,25,26,27,28]
print (age_list)
```

```python
name_list= ["Anu","Meera","Diya","Riya","Kavya"]
print (name_list)
```

## List Operations/Modifications

```python
name_list.append("Yazhini")
print (name_list)
```

```python
age_list.insert(2, 30)
print (age_list)
```

```python
name_list.remove("Yazhini")
print (name_list)
```

```python
age_list.pop()
print (age_list)
```

```python
age_list.extend([29,30,26])
print (age_list)
```

```python
age_list.sort(reverse=True)
print(age_list)
```

```python
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

## Accessing List Elements

```python
print(name_list[-1])
```

```python
print(name_list[2:5])
```

```python
print(name_list[::-1])
```

## Dictionary

```python
student_marks={"Anu": 85,"Meera": 78,"Diya": 92,"Riya": 88,"Kavya": 75}
print (student_marks)
```

```python
print("Diya's mark:", student_marks["Diya"])
```

```python
student_marks["Janani"] = 80
print (student_marks)
```

```python
student_marks["Meera"] = 82
print (student_marks)
```

```python
print("keys:", student_marks.keys())
print("Values:", student_marks.values())
print("key-value pairs:", student_marks.items())
```

## Sets

```python
my_set= {'a','e','i','o','u','a','a','i'}
print (my_set)
```

```python
# The duplicate values 'a' and 'i' are removed because a set does not allow duplicate elements.
# The order of elements in a set may not fixed.
```

```python
my_set[4] = 's'
```

```python
# This is because sets are unordered and do not supporting indexing.
# Therefore, we cannot acess or change a set element using an index.
```

```python
set1 = {1,3,5,7,9}
set2 = {2,3,5,8,10}
print("set 1:", set1)
print("set 2:", set2)
```

```python
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

## Operators & Conditional Statements

```python
score=int(input("Enter your score(0 to 10): "))
if score > 7:
    print("Above Average: Excellent performance! Keep up the good work.")
elif score >= 4 and score <= 7:
    print("Average: Good effort! Keep practicing, there's a room for improvement.")
else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
